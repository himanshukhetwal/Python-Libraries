# Loading the libraries and dataset

```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
# to ignore the unwanted warnings 
import warnings 
warnings.filterwarnings("ignore")


df = pd.read_csv("IPL.csv")
df.head()
```

# Some basic questions

```python
# Which team won the most matches 
match_win = df['match_winner'].value_counts()
sns.barplot(y = match_win.index, x = match_win.values,palette='viridis')
plt.title("Most match wins by team")
```

```python
# Toss decesion trend
sns.countplot(x=df['toss_decision'],palette='rainbow')
plt.title("toss decision trend")
```

```python
# toss winner vs match winner
count = df[df['toss_winner']==df['match_winner']]['match_id'].count()
percentage = (count * 100)/df.shape[0]
percentage.round(2)
```

```python
# how do teams win ? (runs vs wickets)
sns.countplot(x=df['won_by'])
```

# key player performances

```python
# most "player of the match" awards
count = df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(x = count.values, y = count.index,palette='mako')
plt.title("Man of the match")
```

```python
# top 2 scorers
#-> when you have to work with two or more cols you use group by 

high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind='bar')
```

```python
# 10 best bowling figures 
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df # created a new col for highest wickets and will cahnge its datatype to int from string
df['highest_wickets']=df['highest_wickets'].astype(int)
```

```python
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers
```

```python
top_bowlers.plot(kind='bar')
```

# Venue analysis

```python
# most matches played in which venue 
venue = df['venue'].value_counts()
venue
```

```python
sns.barplot(y=venue.index , x = venue.values, palette='rainbow')
```

# custom questions and insights

```python
# who won the highest margin by runs 
df[df['won_by']=='Runs'].sort_values(by = 'margin',ascending=False).head(1)[['match_winner','margin']]
```

```python
# which player had the highest individual score 

df[df['highscore']==df['highscore'].max()][['top_scorer','highscore']]
```

```python
# which bowler has the best bowling figures 

df[df['highest_wickets']==df['highest_wickets'].max()][['best_bowling','best_bowling_figure']]
```
