```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

# MATPLOTLIB BASICS

```python
x = np.linspace(0,5,11)
x
```

```python
y=x**2
y
```

```python
# When you create a plot you have 2 things X and Y 
# Vertical line is Y and horizontal line is Y
plt.plot(x,y)
```

```python
# Now adding some labels on graph
plt.title('Hello')
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
```

# SUBPLOTS IN MATPLOTLIB

```python
# 4X4 subplots
plt.subplot(2,2,1) # 3 args are req first tow are for 2 rows and 2 cols 
#and 3rd arg for selecting which part you want to access
plt.plot(x**2,y*2)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
plt.subplot(2,2,4)
#for creating thins on plot you have to mention just below your code 
plt.plot(x,y) # curve was created in 4th graph as code is written just below the 4th plot
#white background is known as canvas
```

# OBJECT ORIENTED WAY 
used to access the canvas and we can do multiple stuff with it
Matplotlib allows the aspect ratio , DPI and figure size to be specified when the figure object is cretaed.You can use the figsize and dpi keyword arguements.abs
Figsize is a tuple of the width and height of the figure in inches dpi is the dots per inch(pixel per inch )

```python
fig = plt.figure(figsize=(20,8),dpi=500)
axis1 = fig.add_axes([0.1,0.1,1,1])# dist from left , dist from bottom , 80% width and 80% height 
#these are 4 args in this fn
axis1.plot(x,y)
axis1

axis2 = fig.add_axes([0.3,0.3,.4,.4])
axis2.plot(x,y)
# this is how we can make a overlaping plot 
#this is how we can control our canvas 
```

# TYPES OF PLOT

```python
#histogram plot 
plt.hist(x,y)
```

# SAVING PLOT(saving a diagram or graph)

```python
fig=plt.figure()
ax=fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
